# Credit Card Fraud: Exploratory Data Analysis

Data exploration for the ULB Credit Card Fraud Detection dataset from OpenML. This notebook fetches the data, inspects its structure, and prepares it for modelling.

## Download

```python
import os

import pandas as pd
from sklearn.datasets import fetch_openml
```

```python
fraud = fetch_openml(data_id=1597, as_frame=True)

df_fraud = fraud.data
df_fraud["Class"] = fraud.target.astype(int)

os.makedirs("../data/raw", exist_ok=True)
df_fraud.to_csv("../data/raw/creditcard_fraud.csv", index=False)

print("Downloaded and saved:", df_fraud.shape)
```

## Basic structure

284k transactions, 30 columns. All features are numeric: 28 PCA components (V1 through V28), Time, Amount, and the Class label. No categoricals anywhere, which simplifies the pipeline later.

```python
df_fraud = pd.read_csv("../data/raw/creditcard_fraud.csv")

print("Shape:", df_fraud.shape)
print(df_fraud.dtypes.value_counts())
```

## Class distribution

Fraud cases are extremely rare. The dataset has only 492 positives out of 284 807 transactions. At 0.17 percent, this is almost a needle in a haystack and changes which metrics are useful later.

```python
print(df_fraud["Class"].value_counts())
print()
print(df_fraud["Class"].value_counts(normalize=True))
```
